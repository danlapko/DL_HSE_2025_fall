```python
from PIL import Image
import os
from pathlib import Path

def convert_webp_to_png(input_path, output_path=None):
    try:
        with Image.open(input_path) as img:
            if output_path is None:
                output_path = str(Path(input_path).with_suffix('.png'))
            
            if img.mode in ('RGBA', 'LA', 'P'):
                background = Image.new('RGB', img.size, (255, 255, 255))
                if img.mode == 'P':
                    img = img.convert('RGBA')
                background.paste(img, mask=img.split()[-1] if img.mode == 'RGBA' else None)
                img = background
            elif img.mode != 'RGB':
                img = img.convert('RGB')
            
            img.save(output_path, 'PNG', optimize=True)
            print(f"Конвертировано: {input_path} → {output_path}")
            return output_path
    
    except Exception as e:
        print(f"Ошибка при конвертации {input_path}: {e}")
        return None


input_file = "your_dataset/numsiri_fox-0.webp"
convert_webp_to_png(input_file)
```

```python
import google.generativeai as genai
import PIL.Image

def generate_with_gemini(image_path, api_key, keyword="cute_fox_sticker"):
    genai.configure(api_key=api_key)
   
    model = genai.GenerativeModel('gemini-2.5-flash')  
    
    img = PIL.Image.open(image_path)
    
    prompt = f"""Analyze this sticker image and create a descriptive caption in English.
    
    REQUIREMENTS:
    1. MUST include the keyword: '{keyword}' at the end
    2. Maximum 77 tokens (keep it concise)
    3. Describe the character/object, its expression, and style
    4. Mention if it's cartoon, anime, minimalist, kawaii, etc.
    5. Keep it simple but descriptive
    
    Good examples:
    - "A sleepy fox curled up on a cloud, anime style with soft colors, {keyword}"
    - "Happy fox waving with big eyes, cartoon style on white background, {keyword}"
    - "Minimalist fox eating carrot, simple vector art, {keyword}"
    
    Bad examples (too long or missing keyword):
    - "This is a picture of a fox that looks cute"
    - "Animal sticker"
    
    Now describe this sticker image:"""
    
    response = model.generate_content([prompt, img])
    
    description = response.text.strip()
    
    if keyword not in description:
        description = f"{description}, {keyword}"
    
    return description
```
